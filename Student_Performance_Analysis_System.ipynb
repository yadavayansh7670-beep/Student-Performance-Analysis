```python

import numpy as np

print("STUDENT PERFORMANCE ANALYSIS SYSTEM")

names = np.array(["Rohan", "mohan", "shyam", "tushar", "sami",
                  "ayansh", "harinam", "yadav", "gaurav", "aryan"])

subject = np.array(["maths", "physics", "chemistry", "biology", "history"])

marks = np.array([
    [50, 60, 70, 80, 90],
    [54, 61, 78, 84, 97],
    [58, 63, 74, 87, 92],
    [57, 64, 72, 82, 96],
    [54, 66, 73, 82, 94],
    [59, 68, 79, 85, 91],
    [56, 69, 73, 87, 91],
    [58, 65, 75, 87, 94],
    [59, 68, 78, 83, 97],
    [56, 64, 79, 88, 98]
])

s = marks.shape
l = len(names)
n = len(subject)

print("\nDataset Shape:", s)
print("Total Students:", l)
print("Total Subjects:", n)

std_mrk = np.sum(marks, axis=1)
average = np.mean(marks, axis=1)
overall_avg = np.mean(marks)

highest = np.max(marks)
lowest = np.min(marks)

hig_sub = np.max(marks, axis=0)
low_sub = np.min(marks, axis=0)

hig_std = np.max(marks, axis=1)
low_std = np.min(marks, axis=1)

topper = np.argmax(std_mrk)

print("\nOverall Topper:", names[topper])

max_marks = np.max(std_mrk)
perc = max_marks / 500 * 100

print("Topper Marks:", max_marks)
print("Topper Percentage:", perc)

a = np.argsort(std_mrk)[::-1]
top_3 = a[:3]

print("\nTOP 3 STUDENTS")

for i in range(len(top_3)):
    idx = top_3[i]
    print(i + 1, names[idx], std_mrk[idx])

result = np.all(marks >= 40, axis=1)

pass_students = np.where(result)[0]
fail_students = np.where(~result)[0]

print("\nPASS STUDENTS:", names[pass_students])
print("FAIL STUDENTS:", names[fail_students])

print("Total Passed:", len(pass_students))
print("Total Failed:", len(fail_students))

t = np.argmax(marks, axis=0)

topper_sub = names[t]
max_sub_marks = np.max(marks, axis=0)

print("\nSUBJECT WISE TOPPER")

for i in range(len(subject)):
    print(subject[i], topper_sub[i], max_sub_marks[i])

std_perc = (std_mrk / 500) * 100

rank = np.argsort(std_mrk)[::-1]

print("\n========== FINAL STUDENT REPORT ==========")

for i in range(len(names)):
    idx = rank[i]

    print("\nRank:", i + 1)
    print("Name:", names[idx])
    print("Total Marks:", std_mrk[idx])
    print("Average Marks:", average[idx])
    print("Percentage:", std_perc[idx])
    print("Status:", "Pass" if result[idx] else "Fail")

print("\n========== OVERALL STATISTICS ==========")

print("Highest Marks:", highest)
print("Lowest Marks:", lowest)
print("Overall Average:", overall_avg)

print("\nSubject Wise Highest Marks:", hig_sub)
print("Subject Wise Lowest Marks:", low_sub)

print("\nPROJECT COMPLETED SUCCESSFULLY!")
```
